```python
df_orderdetails = pd.read_csv('2021_OrderDetails.csv')
df_orderdetails.info()
df_orderdetails.head()
```

```python
df_orderdetails['MONTH'] = df_orderdetails['ORDER DATE'].astype('datetime64[ns]').dt.month
df_orderdetails
```

```python
#1. Tháng nào có doanh số tốt nhất? Doanh số tháng đó là bao nhiêu?

df_orderdetails.groupby('MONTH').sum()['ORDER VALUE']
```

```python
ORDER_VALUE = df_orderdetails.groupby('MONTH').sum()['ORDER VALUE']
ORDER_VALUE.max()
```

```python
#2. Quận/Huyện nào có doanh số cao nhất? Doanh số đó là bao nhiêu?

ORDER_VALUE_DISTRICTS = df_orderdetails.groupby('OUTLET MASTER REGION').sum()['ORDER VALUE']
DISTRICTS = [DISTRICTS for DISTRICTS, value in ORDER_VALUE_DISTRICTS.items()]
plt.bar(x=DISTRICTS,height=ORDER_VALUE_DISTRICTS)
plt.xticks(DISTRICTS, rotation =90, size =8)
plt.xlabel('DISTRICTS')
plt.ylabel('ORDER VALUE IN VND')
plt.show()
```

```python
ORDER_VALUE_DISTRICT = df_orderdetails.groupby('OUTLET MASTER REGION').sum()['ORDER VALUE']
ORDER_VALUE_DISTRICT.max()
```

```python
#3. Doanh nghiệp cần chạy CTKM vào khung thời gian nào để tăng khả năng mua hàng của khách hàng?

df_orderdetails['HOURS'] = df_orderdetails['ORDER DATE'].astype('datetime64[ns]').dt.hour
ORDER_VALUE_HOURS = df_orderdetails.groupby('HOURS').count()['ORDER ID']
HOURS = [HOURS for HOURS, value in ORDER_VALUE_HOURS.items()]
plt.plot(HOURS,ORDER_VALUE_HOURS)
plt.grid()
plt.xticks(HOURS, rotation =90, size =8)
plt.xlabel('HOURS')
plt.ylabel('ORDER NUMBER')
plt.show()
```
